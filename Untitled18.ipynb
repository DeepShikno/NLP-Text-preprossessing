```python
from nltk.stem import PorterStemmer, LancasterStemmer, SnowballStemmer
from nltk.tokenize import word_tokenize

def compare_stemmers(text):
    porter = PorterStemmer()
    lancaster = LancasterStemmer()
    snowball = SnowballStemmer("english")
    tokens = word_tokenize(text)

    print(f"{'Word':<15}{'Porter':<15}{'Lancaster':<15}{'Snowball':<15}")
    print("-" * 60)
    for word in tokens:
        print(f"{word:<15}{porter.stem(word):<15}{lancaster.stem(word):<15}{snowball.stem(word):<15}")

sample_text = "The runners were running rapidly and easily during the championship races"
compare_stemmers(sample_text)
```

```python
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer

def preprocess_pipeline(text):
    stop_words = set(stopwords.words("english"))
    stemmer = SnowballStemmer("english")

    tokens = word_tokenize(text)
    print("1. Tokens:", tokens)

    tokens = [t.lower() for t in tokens if t not in string.punctuation]
    print("2. Lowercased & punctuation removed:", tokens)

    filtered_tokens = [t for t in tokens if t not in stop_words]
    print("3. Stop words removed:", filtered_tokens)

    stemmed_tokens = [stemmer.stem(t) for t in filtered_tokens]
    print("4. Stemmed tokens:", stemmed_tokens)
    return stemmed_tokens

raw_sentence = "The children were playing happily in the gardens while the birds were singing loudly."
preprocess_pipeline(raw_sentence)
```

```python
from nltk.tokenize import TweetTokenizer
from nltk.stem import SnowballStemmer, WordNetLemmatizer

tweet = "The new phone cameras are AMAZING!! #Loved!! @TechReview"


tknzr = TweetTokenizer(preserve_case=True, reduce_len=True, strip_handles=False)
tokens = tknzr.tokenize(tweet)
print("1. Tokens:", tokens)

stemmer = SnowballStemmer("english")
stemmed = [stemmer.stem(t) for t in tokens]
print("2. Stemmed:", stemmed)


lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(t.lower(), pos="v") for t in tokens]
print("3. Lemmatized:", lemmatized)
```
